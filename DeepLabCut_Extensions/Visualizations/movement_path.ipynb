```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the CSV
#Change the header column based on which coordinate you want to calculate the path from
df = pd.read_csv('path/to/csv',header=[1,2])


# Check the first few rows to confirm it loaded correctly
print(df.head())

# Plotting the path
plt.figure(figsize=(8, 6))
plt.plot(df['x'], df['y'], marker='o', linestyle='-')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.title('Mouse Movement Path')
plt.gca().invert_yaxis()  # Invert y-axis if needed (like in images/videos)
plt.grid(True)
plt.show()
```
