```python
import pandas as pd
import matplotlib.pyplot as plt

# --- Load your tracking data
tracking_df = pd.read_csv('path/to/csv')

# --- Create a Behavior column based on your rules
def determine_behavior(row):
    nose_speed = row['Nose_Speed']
    tailbase_speed = row['Tailbase_Speed']

    if pd.notna(nose_speed) and pd.notna(tailbase_speed):
        if nose_speed > 10 and tailbase_speed < 10:
            return 'Scanning'
        else:
            return 'Other'
    else:
        return 'Other'

tracking_df['Behavior'] = tracking_df.apply(determine_behavior, axis=1)

# --- Plot locations when mouse is scanning
mask_scanning = tracking_df['Behavior'] == 'Scanning'

nose_x = tracking_df.loc[mask_scanning, 'Nose']
nose_y = tracking_df.loc[mask_scanning, 'Nose.1']
tail_x = tracking_df.loc[mask_scanning, 'Tail_base']
tail_y = tracking_df.loc[mask_scanning, 'Tail_base.1']

plt.figure(figsize=(8, 6))
plt.scatter(nose_x, nose_y, label='Nose', alpha=0.7, s=10)
plt.scatter(tail_x, tail_y, label='Tail_base', alpha=0.7, s=10)
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Mouse Positions During Scanning')
plt.legend()
plt.gca().invert_yaxis()  # Important for video coordinate systems
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#adjust until you get a good visual! you can change the size of the CSV is you need less examples to show the relationship
# --- Load your tracking data
tracking_df = pd.read_csv('/path/to/csv')

# --- Define the behavior based on scanning or rearing
def determine_behavior(row):
    nose_speed = row['Nose_Speed']
    tailbase_speed = row['Tailbase_Speed']
    nose_x = row['Nose']

    if pd.notna(nose_speed) and pd.notna(tailbase_speed) and pd.notna(nose_x):
        # First check for scanning
        if nose_speed > 10 and tailbase_speed < 10:
            return 'Scanning'
        # Then check for rearing
        elif nose_x < 53 or nose_x > 300:
            return 'Rearing'
        else:
            return 'Other'
    else:
        return 'Other'

# Apply the behavior labeling
tracking_df['Behavior'] = tracking_df.apply(determine_behavior, axis=1)

# --- Plot: One plot per behavior (Scanning and Rearing)

for behavior in ['Scanning', 'Rearing']:
    mask = tracking_df['Behavior'] == behavior

    nose_x = tracking_df.loc[mask, 'Nose']
    nose_y = tracking_df.loc[mask, 'Nose.1']
    tail_x = tracking_df.loc[mask, 'Tail_base']
    tail_y = tracking_df.loc[mask, 'Tail_base.1']

    plt.figure(figsize=(8, 6))
    plt.scatter(nose_x, nose_y, label='Nose', alpha=0.7, s=10)
    plt.scatter(tail_x, tail_y, label='Tail_base', alpha=0.7, s=10)
    plt.xlabel('X Position')
    plt.ylabel('Y Position')
    plt.title(f'Mouse Positions During {behavior}')
    plt.legend()
    plt.gca().invert_yaxis()  # To match video frame orientation
    plt.grid(True)
    plt.show()
```
